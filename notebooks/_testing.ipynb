```python
import pysolr

solrCoreUrl = "http://localhost:8983/solr/my1stcore"

def runDiagnostics():
    solr = pysolr.Solr(solrCoreUrl, timeout=30)
    
    # Test 1: Query string in q parameter
    res1 = solr.search("language:de AND type:Text", rows=0)
    print(f"Test 1 (q parameter hits): {res1.hits}")
    
    # Test 2: Filter query in fq parameter
    res2 = solr.search("*:*", fq=["language:de", "type:Text"], rows=0)
    print(f"Test 2 (fq parameter hits): {res2.hits}")
    
    # Test 3: Pull a few documents from the 31k set and verify them
    res3 = solr.search("language:de AND type:Text", rows=10, fl="id,language,type")
    print("\n--- Spot-checking first 10 documents from query set ---")
    for doc in res3:
        docId = doc.get("id")
        docLang = doc.get("language")
        docType = doc.get("type")
        print(f"ID: {docId} | language: {docLang} | type: {docType}")

if __name__ == "__main__":
    runDiagnostics()
```

```python
import pysolr
from pathlib import Path

repoP = Path.cwd().parent

solrCoreUrl = "http://localhost:8983/solr/my1stcore"
outputFilePath = repoP / "data/assignment-data/ebook_ids.txt"

def getEbookIds():
    solr = pysolr.Solr(solrCoreUrl, timeout=30)
    
    query = "language:de AND type:Text"
    
    # PySolr handles deep paging automatically via cursorMark='*'
    results = solr.search(
        query,
        sort="id asc",
        fl="id",
        cursorMark="*"
    )
    
    # Extract IDs directly from the auto-paginating iterator
    allIds = [doc.get("id") for doc in results if "id" in doc]

    # Write unique IDs to text file
    with open(outputFilePath, "w", encoding="utf-8") as f:
        for ebookId in allIds:
            f.write(f"{ebookId}\n")
            
    print(f"Anzahl gespeicherter E-Book-IDs: {len(allIds)}")

if __name__ == "__main__":
    getEbookIds()
```

```python
# this isn't working properly

import pysolr
from pathlib import Path

repoP = Path.cwd().parent.parent

solrCoreUrl = "http://localhost:8983/solr/my1stcore"
outputFilePath = repoP / "data/assignment-data/ebook_ids.txt"

def getEbookIds():
    
    # connect to solr core
    solr = pysolr.Solr(solrCoreUrl, timeout=30)
    
    # define query for german language and text document type
    query = "language:de AND type:Text"
    cursorMark = "*"        # tell solr start pagination at very beginning
    allIds = []
    
    while True:

        searchParams = {
            "q": query,
            "cursorMark": cursorMark,
            "sort": "id asc",
            "rows": 100,
            "fl": "id",
        }

        
        # fetch batch using cursor mark
        results = solr.search(**searchParams)
        
        # extract ids from current batch
        batchIds = [doc.get("id") for doc in results if "id" in doc]
        if not batchIds:
            break
            
        allIds.extend(batchIds)
        
        # get next cursor position
        nextCursor = results.nextCursorMark
        if cursorMark == nextCursor:            # if reached last page of results -> cursorMark identical to last cursorMark 
            break
        cursorMark = nextCursor

    # write ids to text file
    with open(outputFilePath, "w", encoding="utf-8") as f:
        for ebookId in allIds:
            f.write(f"{ebookId}\n")
            
    print(f"Anzahl gespeicherter E-Book-IDs: {len(allIds)}")

if __name__ == "__main__":
    getEbookIds()
```
